```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("pca.dat")
frame = data[:, 0]
pc1 = data[:, 1]
pc2 = data[:, 2]
```

```python
trajs = [data[n*1000:(n+1)*1000] for n in range (0,8)]
```

```python
fig, axs = plt.subplots(4, 2,figsize=(12,20))
fig.suptitle("PCA Trajectories Across Seeds", fontsize=18)
fig.supxlabel("PC1")
fig.supylabel("PC2")
i = 0
for m in range(0,4):
    for n in range(0,2):
        frame = trajs[i][:, 0]
        time = frame*0.002 - 2*i
        pc1 = trajs[i][:, 1]
        pc2 = trajs[i][:, 2]
        sc = axs[m,n].scatter(pc1, pc2, c=time, cmap='viridis', s=20, edgecolor='none')
        cbar = plt.colorbar(sc)
        cbar.set_label("Simulation Time (ns)", fontsize=12)
        axs[m,n].set_title(f'Seed {i}')
        i += 1
```

```python

# Normalize time for coloring
time = frame*0.002

plt.figure(figsize=(6,5))

sc = plt.scatter(pc1, pc2, c=time, cmap='viridis',
                 s=20, edgecolor='none')

cbar = plt.colorbar(sc)
cbar.set_label("Simulation Time (ns)", fontsize=12)

plt.xlabel("PC1 (Å)", fontsize=14)
plt.ylabel("PC2 (Å)", fontsize=14)
plt.title("PCA Projection: PC1 vs PC2", fontsize=14)

plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

plt.tight_layout()
```

```python

from matplotlib import pyplot as plt
from scipy.stats import gaussian_kde
import sys

xy = np.vstack([pc1,pc2])
z = gaussian_kde(xy)(xy)

plt.figure(figsize=(6,5))
scc = plt.scatter(pc1,pc2, c=z, s=5)

cbar = plt.colorbar(scc)
cbar.set_label("Gaussian kde", fontsize=12)

plt.xlabel('pc1',fontsize=14)
plt.ylabel('pc2', fontsize=14)
plt.title("toi day", fontsize=20)

plt.show()
```
